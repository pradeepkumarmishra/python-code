```python
df1['income_level'] = df1['income_level'].map({'<10L' : 1, '10L - 25L' : 2, '25L - 40L' : 3, '> 40L': 4})
```

```python
df1['income_level'].unique()
```

```python
df1['insurance_plan'] = df1['insurance_plan'].map({'Bronze' : 1, 'Silver':2, 'Gold': 3})
```

```python
risk_scores = {
    "diabetes": 6,
    "heart disease": 8,
    "high blood pressure":6,
    "thyroid": 5,
    "no disease": 0,
    "none":0
}

df1[['disease1', 'disease2']] = df1['medical_history'].str.split('&', expand=True).apply(lambda x : x.str.lower())
df1['disease1'].fillna('none', inplace = True)
df1['disease2'].fillna('none', inplace = True)
df1[['disease1', 'disease2']] = df1[['disease1', 'disease2']].apply(lambda x : x.map(risk_scores))
df1['total_score'] = 0
for score in ['disease1', 'disease2']:
    df1['total_score'] += df1[score]
```

```python
df1.select_dtypes(include='object').columns
```

```python
df1 = pd.get_dummies(df1, columns=['gender', 'region', 'marital_status', 'bmi_category', 'smoking_status',
       'employment_status'])
```

```python
df1.sample(3)
```

```python
df1 = df1.drop(axis = 1, columns= ['medical_history','disease1', 'disease2', 'total_score'])
```

```python
df1.corr()
```

```python
plt.figure(figsize=(20,12))
sns.heatmap(df1.corr(), annot= True,)
```

```python
df1[['age', 'number_of_dependants', 'income_level', 'income_lakhs',
       'insurance_plan', 'annual_premium_amount']]
```

```python
X = df1.drop(axis = 1, columns= ['annual_premium_amount'])
y = df1['annual_premium_amount']

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
columns_to_scale = ['age', 'number_of_dependants', 'income_level', 'income_lakhs','insurance_plan']
X[columns_to_scale] = scaler.fit_transform(X[columns_to_scale])
```

```python
X.describe()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state=42)

lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
y_pred = lr_model.predict(X_test)
lr_model.score(X_train, y_train), lr_model.score(X_test, y_test)
lr_model.coef_, lr_model.intercept_
```

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor()
xgb_model.fit(X_train, y_train)
y_pred_xgb = xgb_model.predict(X_test)
xgb_model.score(X_train, y_train), xgb_model.score(X_test, y_test)
```

```python
from sklearn.model_selection import RandomizedSearchCV
params =  {'n_estimators' : [20, 40, 50],
           'learning_rate': [0.01, 0.1, 0.2],
           'max_depth': [3, 4, 5]
          }
rsv = RandomizedSearchCV(xgb_model, params, n_iter = 10, cv = 3, scoring='r2', random_state=42, n_jobs=-1)
rsv.fit(X_train, y_train)
rsv.best_score_
best_model = rsv.best_estimator_
rsv.best_params_
```

```python

feature_importances = best_model.feature_importances_
coef_df = pd.DataFrame(feature_importances,index = X_train.columns, columns= ['Coefficients'])
coef_df.sort_values(by='Coefficients', ascending=True)

plt.figure(figsize= (8, 4))
plt.barh(coef_df.index, coef_df['Coefficients'], color='steelblue')
plt.xlabel('Coefficient Value')
plt.title('Feature Importance in XGBoost')
plt.show()
```

```python
y_pred = best_model.predict(X_test)
```

```python
residuals = y_pred - y_test
residuals
```

```python
rs_pct = residuals*100 / y_test

results = {
    'actual': y_test,
    'predicted': y_pred,
    'diff':residuals,
    'diff_pct': rs_pct
}
results_df = pd.DataFrame(results)
results_df
```

```python
sns.histplot(results_df['diff_pct'], kde=True)
plt.title('Distribution of Residuals')
plt.xlabel('Diff PCT')
plt.ylabel('Frequency')
plt.show()
```

```python
extrem_results_df = results_df[results_df['diff_pct'] > 10 ]
extrem_results_df
```

```python
extrem_results_df[abs(extrem_results_df.diff_pct) > 50]
```

```python
extrem_results_df = X_test.loc[extrem_results_df.index]

for feature in X_test.columns:
    plt.figure(figsize= (10, 4))
    sns.histplot(extrem_results_df[feature], kde = True)
    sns.histplot(X_test[feature], kde = True)
    plt.show()
```

```python
reversed_df = pd.DataFrame()
reversed_df[columns_to_scale] = scaler.inverse_transform(extrem_results_df[columns_to_scale])
reversed_df
```
